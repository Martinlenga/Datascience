```python
# RDD - lower level distributed data structure in Spark
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName('Practice') \
    .master("local") \
    .getOrCreate()

# create SparkContext - entry point for lower level for rdd operations
sc = spark.sparkContext
num = [2, 4, 7, 9, 13, 5, 11]

num_rdd = sc.parallelize(num)
```

```python
def squares(num):
    return num ** 2

def odd(num):
    return num % 2 == 1

def even(num):
    return num % 2 == 0

# Use RDD methods .map() and .filter() instead of Python built-ins
num_squares = num_rdd.map(squares)
num_even = num_rdd.filter(even)
num_odd = num_rdd.filter(odd)

# Use .collect() to turn the RDD results into a printable list
print(num_squares.collect())
print(num_even.collect())
print(num_odd.collect())
```
