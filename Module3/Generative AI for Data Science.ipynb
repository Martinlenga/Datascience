## Scenario 1: Retail Inventory Analysis

### My Prompt:
“Act as a senior retail data analyst and Python developer. Using pandas, matplotlib and seaborn, generate Python code for a retail inventory analysis system using a DataFrame with the following columns: product_id, category, stock_level, last_restock_date, sales_last_30_days, supplier_lead_time and unit_cost.

The code should:
- Calculate inventory turnover rates.
- Identify slow-moving inventory items.
- Predict potential stockouts using current stock and sales trends.
- Handle date calculations correctly.
- Include data validation and error handling.
- Create meaningful visualizations for inventory insights.
- Include comments and documentation for readability.”

```python
# Generated Code

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Sample validation function
def validate_data(df):
    required_columns = [
        'product_id', 'category', 'stock_level',
        'last_restock_date', 'sales_last_30_days',
        'supplier_lead_time', 'unit_cost'
    ]

    for col in required_columns:
        if col not in df.columns:
            raise ValueError(f"Missing column: {col}")

    df['last_restock_date'] = pd.to_datetime(df['last_restock_date'])

    numeric_cols = [
        'stock_level',
        'sales_last_30_days',
        'supplier_lead_time',
        'unit_cost'
    ]

    for col in numeric_cols:
        if (df[col] < 0).any():
            raise ValueError(f"Negative values found in {col}")

    return df

def calculate_inventory_metrics(df):
    df = validate_data(df)

    # Inventory turnover
    df['inventory_turnover_rate'] = (
        df['sales_last_30_days'] / df['stock_level']
    )

    # Slow moving items
    df['slow_moving'] = df['inventory_turnover_rate'] < 0.5

    # Daily sales estimate
    df['daily_sales'] = df['sales_last_30_days'] / 30

    # Days until stockout
    df['days_until_stockout'] = (
        df['stock_level'] / df['daily_sales']
    )

    # Predict stockout risk
    df['potential_stockout'] = (
        df['days_until_stockout'] < df['supplier_lead_time']
    )

    return df

def create_visualizations(df):
    sns.histplot(df['inventory_turnover_rate'])
    plt.title("Inventory Turnover Distribution")
    plt.show()

    sns.scatterplot(
        data=df,
        x='stock_level',
        y='sales_last_30_days',
        hue='potential_stockout'
    )
    plt.title("Stock vs Sales")
    plt.show()
```

```python
# Final Solution 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def validate_data(df):
    required_columns = [
        'product_id', 'category', 'stock_level',
        'last_restock_date', 'sales_last_30_days',
        'supplier_lead_time', 'unit_cost'
    ]

    missing = [col for col in required_columns if col not in df.columns]

    if missing:
        raise ValueError(f"Missing columns: {missing}")

    df['last_restock_date'] = pd.to_datetime(
        df['last_restock_date'],
        errors='coerce'
    )

    if df['last_restock_date'].isnull().any():
        raise ValueError("Invalid dates detected.")

    numeric_cols = [
        'stock_level',
        'sales_last_30_days',
        'supplier_lead_time',
        'unit_cost'
    ]

    for col in numeric_cols:
        if (df[col] < 0).any():
            raise ValueError(f"Negative values in {col}")

    return df

def inventory_analysis(df):
    df = validate_data(df)

    df['daily_sales'] = df['sales_last_30_days'] / 30

    df['inventory_turnover_rate'] = np.where(
        df['stock_level'] > 0,
        df['sales_last_30_days'] / df['stock_level'],
        0
    )

    df['slow_moving'] = df['inventory_turnover_rate'] < 0.5

    df['days_until_stockout'] = np.where(
        df['daily_sales'] > 0,
        df['stock_level'] / df['daily_sales'],
        np.inf
    )

    df['potential_stockout'] = (
        df['days_until_stockout'] <
        df['supplier_lead_time']
    )

    return df

def create_visualizations(df):
    plt.figure(figsize=(10, 6))
    sns.barplot(
        data=df,
        x='category',
        y='inventory_turnover_rate'
    )
    plt.xticks(rotation=45)
    plt.title("Inventory Turnover by Category")
    plt.show()

    plt.figure(figsize=(10, 6))
    sns.scatterplot(
        data=df,
        x='stock_level',
        y='sales_last_30_days',
        hue='potential_stockout'
    )
    plt.title("Potential Stockout Analysis")
    plt.show()
```

### My Prompt:

“Act as a senior Python debugging expert and data analyst. Review the following engagement analytics code and identify all logic, validation, and calculation issues. Fix bounce rate calculations, prevent impossible values such as bounce rates above 100% or negative session times, implement proper session duration handling, add validation checks, and create summary visualizations.”

```python
# Generated Code

import pandas as pd
import matplotlib.pyplot as plt

def analyze_user_engagement(logs_df):

    # Remove invalid durations
    logs_df = logs_df[logs_df['duration'] >= 0]

    # Bounce rate calculation
    session_summary = logs_df.groupby('session_id').agg({
        'page_views': 'sum',
        'duration': 'sum'
    })

    bounce_rate = (
        (session_summary['page_views'] == 1).mean()
    ) * 100

    avg_session_time = session_summary['duration'].mean()

    pages_per_session = session_summary['page_views'].mean()

    metrics = {
        'bounce_rate': bounce_rate,
        'avg_session_time': avg_session_time,
        'pages_per_session': pages_per_session
    }

    return metrics
```

### Follow-up Prompts:
- “Explain why the original bounce rate logic failed.”
- “Add device-level visualizations.”

- “Prevent negative durations from entering calculations.”

```python
# Final Solution 

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def validate_logs(df):

    required_cols = [
        'session_id',
        'page_views',
        'duration',
        'device_type'
    ]

    for col in required_cols:
        if col not in df.columns:
            raise ValueError(f"Missing column: {col}")

    df = df.dropna(subset=['session_id'])

    df = df[df['duration'] >= 0]

    return df

def analyze_user_engagement(logs_df):

    logs_df = validate_logs(logs_df)

    session_summary = logs_df.groupby('session_id').agg({
        'page_views': 'sum',
        'duration': 'sum'
    })

    bounce_rate = (
        (session_summary['page_views'] == 1).mean()
    ) * 100

    avg_session_time = session_summary['duration'].mean()

    pages_per_session = session_summary['page_views'].mean()

    device_metrics = logs_df.groupby('device_type').agg({
        'session_id': 'nunique',
        'duration': 'mean',
        'page_views': 'mean'
    })

    metrics = {
        'bounce_rate': bounce_rate,
        'avg_session_time': avg_session_time,
        'pages_per_session': pages_per_session
    }

    plt.figure(figsize=(8, 5))
    sns.barplot(
        x=device_metrics.index,
        y=device_metrics['duration']
    )
    plt.title("Average Session Duration by Device")
    plt.show()

    return metrics, device_metrics
```

## Scenario 3: Customer Segmentation Query

### My Prompt:

“Act as a senior SQL data analyst. Generate an optimized SQL query using CTEs to identify active users who logged in within the last 30 days, determine high-value customers in the top 20% by total spending, and analyze user preference trends. Include proper date filtering, percentile calculations, and optimization best practices.”

```python
# Generated Code

generated_query = """
WITH active_users AS (
    SELECT
        ua.user_id,
        ua.last_login_date,
        ua.account_type
    FROM user_activity ua
    WHERE ua.last_login_date >= CURRENT_DATE - INTERVAL '30 days'
),

customer_spending AS (
    SELECT
        t.user_id,
        SUM(t.amount) AS total_spending
    FROM transactions t
    GROUP BY t.user_id
),

high_value_customers AS (
    SELECT
        user_id,
        total_spending
    FROM (
        SELECT
            user_id,
            total_spending,
            NTILE(5) OVER (
                ORDER BY total_spending DESC
            ) AS spending_group
        FROM customer_spending
    ) ranked
    WHERE spending_group = 1
)

SELECT
    au.user_id,
    au.account_type,
    hvc.total_spending,
    up.communication_preference,
    up.interface_theme,
    up.notification_settings
FROM active_users au
JOIN high_value_customers hvc
    ON au.user_id = hvc.user_id
JOIN user_preferences up
    ON au.user_id = up.user_id;
"""

print(generated_query)
```

### Follow-up Prompts:

```python
# Final Solution

final_query = """
WITH active_users AS (
    SELECT
        user_id,
        account_type
    FROM user_activity
    WHERE last_login_date >= CURRENT_DATE - INTERVAL '30 days'
),

customer_spending AS (
    SELECT
        user_id,
        SUM(amount) AS total_spending
    FROM transactions
    GROUP BY user_id
),

ranked_customers AS (
    SELECT
        user_id,
        total_spending,
        NTILE(5) OVER (
            ORDER BY total_spending DESC
        ) AS percentile_group
    FROM customer_spending
),

high_value_customers AS (
    SELECT *
    FROM ranked_customers
    WHERE percentile_group = 1
)

SELECT
    au.user_id,
    au.account_type,
    hvc.total_spending,
    up.communication_preference,
    up.interface_theme,
    up.notification_settings
FROM active_users au
INNER JOIN high_value_customers hvc
    ON au.user_id = hvc.user_id
INNER JOIN user_preferences up
    ON au.user_id = up.user_id;
"""

print(final_query)
```

Percentile calculations were implemented using the NTILE(5) window function, where the top 20% of customers were identified as the first group.

#### What approaches to date filtering were used?
The query used CURRENT_DATE - INTERVAL '30 days' to dynamically identify recently active users.

#### How was the query optimized?
The optimized solution used Common Table Expressions (CTEs) to improve readability and modularity. The prompts also encouraged filtering early and minimizing unnecessary joins.
