```python
# CodeGrade step0

from sklearn.datasets import fetch_california_housing
import pandas as pd
import numpy as np
import statsmodels.api as sm
import os
import tarfile
import joblib

from sklearn.datasets._base import _pkl_filepath, get_data_home

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score

from statsmodels.stats.outliers_influence import variance_inflation_factor

archive_path = "cal_housing.tgz"
data_home = get_data_home(data_home=None)

if not os.path.exists(data_home):
    os.makedirs(data_home)

filepath = _pkl_filepath(data_home, 'cal_housing.pkz')

with tarfile.open(mode="r:gz", name=archive_path) as f:
    cal_housing = np.loadtxt(
        f.extractfile('CaliforniaHousing/cal_housing.data'),
        delimiter=','
    )

    columns_index = [8, 7, 2, 3, 4, 5, 6, 1, 0]
    cal_housing = cal_housing[:, columns_index]

    joblib.dump(cal_housing, filepath, compress=6)

california = fetch_california_housing(as_frame=True)

data = california.data
data['MedianHouseValue'] = california.target
```

```python
# CodeGrade step1

threshold = data['MedianHouseValue'].median()

data['HighValue'] = (
    data['MedianHouseValue'] > threshold
).astype(int)

data['HighValue'].unique()
```

```python
# CodeGrade step2

X = data[['MedInc', 'AveRooms', 'AveOccup']]

y = data['HighValue']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42
)

X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
# CodeGrade step3

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

X_test_scaled.shape
```

```python
# CodeGrade step4

model = LogisticRegression(random_state=42)

model.fit(
    X_train_scaled,
    y_train
)

model.intercept_
```

```python
# CodeGrade step5

model.coef_
```

```python
# CodeGrade step6

y_pred_prob = model.predict_proba(X_test_scaled)[:, 1]

y_pred_class = model.predict(X_test_scaled)

y_pred_prob[:5], y_pred_class[:5]
```

```python
# CodeGrade step7

confusion_matrix(
    y_test,
    y_pred_class
)
```

```python
# CodeGrade step8

round(
    accuracy_score(
        y_test,
        y_pred_class
    ),
    4
)
```

```python
# CodeGrade step9

vifs = []

for i in range(X_train_scaled.shape[1]):
    vifs.append(
        round(
            variance_inflation_factor(
                X_train_scaled,
                i
            ),
            3
        )
    )

vifs
```
