```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.utils import resample
```

```python
data = pd.DataFrame({
    "hours_studied": [1,2,3,5,6,7,8,9],
    "final_score": [22,34,45,15,59,63,85,77],
})
data
```

```python
X,y = data[["hours_studied"]], data.final_score
model = LinearRegression()

model.fit(X,y)

X_test = pd.DataFrame({"hours_studied": [10,7, 11]})
model.predict(X_test)
```
